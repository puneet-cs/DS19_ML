# Random forest Classifier

```python
# Libraries
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import pandas as pd
```

```python
# import the dataset
df = pd.read_csv("driver_accident_risk_dataset.csv")
df
```

```python
df.accident_risk.unique()
```

```python
x = df.drop("accident_risk", axis = 1)
y = df.accident_risk
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
rf = RandomForestClassifier(n_estimators = 100, n_jobs = -1, max_samples = 0.8, max_features = 0.5)
```

```python
rf.fit(x_train_scaled, y_train)
```

```python
rf.score(x_test_scaled, y_test)
```

# Random forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
df = pd.read_csv("housing_data_50k_extended.csv")
df
```

```python
x = df.drop("price", axis =1)
y = df.price 
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size = 0.2, random_state = 40)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
rfr = RandomForestRegressor(n_estimators = 50, n_jobs = -1, max_samples = 0.7, max_features = 0.6)
rfr.fit(x_train_scaled, y_train)
```

```python
rfr.score(x_test_scaled, y_test)
```

# Bagging 

```python
from sklearn.ensemble import BaggingClassifier
from sklearn.neighbors import KNeighborsClassifier
```

```python
df = pd.read_csv("driver_accident_risk_dataset.csv")
df
```

```python
x = df.drop("accident_risk", axis = 1)
y = df.accident_risk
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
base_model = KNeighborsClassifier(n_neighbors = 5)

bagging_model = BaggingClassifier(estimator=base_model, n_estimators = 20, max_samples = 0.8)

bagging_model.fit(x_train_scaled, y_train)
```

```python
bagging_model.score(x_test_scaled, y_test)
```
